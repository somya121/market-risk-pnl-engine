```python
from pathlib import Path
import pandas as pd
import sys
PROJECT_ROOT = Path.cwd().parent
sys.path.insert(0,str(PROJECT_ROOT),)
risk_metrics = pd.read_csv(
    PROJECT_ROOT
    / "reports"
    / "risk_metrics.csv"
)

stress_results = pd.read_csv(
    PROJECT_ROOT
    / "reports"
    / "stress_results.csv"
)

daily_pnl = pd.read_csv(
    PROJECT_ROOT
    / "reports"
    / "daily_pnl.csv"
)
```

```python
var_99 = float(
    risk_metrics.loc[
        risk_metrics["metric"]
        == "Historical VaR 99%",
        "value",
    ].iloc[0]
)

es_99 = float(
    risk_metrics.loc[
        risk_metrics["metric"]
        == "Expected Shortfall 99%",
        "value",
    ].iloc[0]
)

daily_loss = daily_pnl[
    "daily_pnl"
].min()

stress_loss = stress_results[
    "stress_pnl"
].min()
```

```python
from src.risk.limit_monitor import (
    evaluate_risk_limits,
)

limit_results = evaluate_risk_limits(
    var_99=var_99,
    expected_shortfall_99=es_99,
    daily_loss=daily_loss,
    stress_loss=stress_loss,
)

limit_results
```

```python
limit_results.to_csv(
    PROJECT_ROOT
    / "reports"
    / "risk_limit_results.csv",
    index=False,
)
```

```python
breaches = limit_results[
    limit_results["status"]
    == "BREACH"
]

warnings = limit_results[
    limit_results["status"]
    == "WARNING"
]

print("BREACHES")
display(breaches)

print("WARNINGS")
display(warnings)
```
